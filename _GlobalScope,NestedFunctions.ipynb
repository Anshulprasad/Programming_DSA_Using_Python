scope of a name

```python
def f():
    y = x
    print(y)

x = 7
f()
```

global names that point to immutable values can't be updated within a function.
because they then become local names

```python
def f():
    y = x
    print(y)
    x = 22

x = 7
f()
```

global names that point to mutable values can be updated within a function.

```python
def f():
    y = x[0]
    print(y)
    x[0] = 22

x = [7]
f()
```

```python
def initialize(N):
    for key in ['queen','row','col','nwtose','swtone']:
        board[key] = {}
    # board = {'queen':{},'row':{},'col':{},'nwtose':{},'swtone':{}}

    for i in range(N):
        board['queen'][i] = -1
        board['row'][i] = 0
        board['col'][i] = 0

    for i in range(-(N-1),N):
        board['nwtose'][i] = 0

    for i in range(2*N-1):
        board['swtone'][i] = 0

def printboard():
    for row in sorted(board['queen'].keys()):
        print((row,board['queen'][row]),end=' ')
    print()

def free(i,j):
    return board['row'][i] == board['col'][j] == board['nwtose'][j-i] == board['swtone'][j+i] == 0

def addqueen(i,j):
    board['queen'][i] = j
    board['row'][i] = 1
    board['col'][j] = 1
    board['nwtose'][j-i] = 1
    board['swtone'][j+i] = 1

def undoqueen(i,j):
    board['queen'][i] = -1
    board['row'][i] = 0
    board['col'][j] = 0
    board['nwtose'][j-i] = 0
    board['swtone'][j+i] = 0

def placeQueen(i):
    for j in range(N):
        
        if free(i,j):
            addqueen(i,j)

            if i == N-1:
                return True
            else:
                extendsoln = placeQueen(i+1)
        
            if extendsoln:
                return True
            else:
                undoqueen(i,j)

    else:
        return False


N = int(input('How many queens?\n'))
board = {}
initialize(N)
if placeQueen(0):
    printboard()
```

Global immutable values

```python
def f():
    global x
    y = x
    print(y)
    x = 22

x = 7
f()
print(x)
```

Nest function definitions

```python
def f():
    def g(a):
        return a+1
    
    def h(b):
        return 2*b
    
    global x
    y = g(x) + h(x)
    print(y)
    x = 22

x = 6
f()
print(x)
```
