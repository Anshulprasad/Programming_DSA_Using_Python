search for value in a list

```python
def findpos(l,v):

    found,i = False,0
    while i<len(l):
        if not found and v == l[i]:
            found,pos = True,i
        i = i+1

    if not found:
        pos = -1
    return pos

print(findpos([1,2,3,4],3))
```

```python
def findpos(l,v):

    pos,i= -1,0

    for x in l:
        if v == x:
            pos = i
            break
        i = i+1

    return pos

print(findpos([1,2,3,4],3))
```

```python
def findpos(l,v):

    pos = -1

    for i in range(len(l)):
        if v == l[i]:
            pos = i
            break

    return pos

print(findpos([1,2,3,4],3))
```
