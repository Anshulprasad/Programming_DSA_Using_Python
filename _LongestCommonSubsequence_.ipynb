Longest Common Subword

Dynamic Programming
- O(n^2)

```python
def func_LCW(u,v):
    LCW = [[c for c in range(len(v)+1)] for r in range(len(u)+1)]

    for r in range(len(u)+1):
        LCW[r][len(v)] = 0

    for c in range(len(v)+1):
        LCW[len(u)][c] = 0
    
    maxLCW = 0

    for c in range(len(v)-1,-1,-1):
        for r in range(len(u)-1,-1,-1):

            if u[r] == v[c]:
                LCW[r][c] = 1 + LCW[r+1][c+1]
            else:
                LCW[r][c] = 0
            
            if LCW[r][c] > maxLCW:
                maxLCW = LCW[r][c]
    
    return maxLCW

print(func_LCW('bisect','secret'))
```

Longest Common Subsequence

```python
def func_LCS(u,v):
    LCS = [[c for c in range(len(v)+1)] for r in range(len(u)+1)]

    for r in range(len(u)+1):
        LCS[r][len(v)] = 0

    for c in range(len(v)+1):
        LCS[len(u)][c] = 0
    
    for c in range(len(v)-1,-1,-1):
        for r in range(len(u)-1,-1,-1):

            if u[r] == v[c]:
                LCS[r][c] = 1 + LCS[r+1][c+1]
            else:
                LCS[r][c] = max(LCS[r+1][c],LCS[r][c+1])
    
    return LCS[0][0]

print(func_LCS('bisect','secret'))
```
