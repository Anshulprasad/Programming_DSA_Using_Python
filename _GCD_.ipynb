#### algorithms and programming: simple gcd

time complexity is propotional to m and n

```python
def gcd(m,n):

    fm = []
    for i in range(1,m+1):
        if m%i == 0:
            fm.append(i)
    
    fn = []
    for j in range(1,n+1):
        if n%j == 0:
            fn.append(j)
    
    cf = []
    for f in fm:
        if f in fn:
            cf.append(f)
    
    return cf[-1]

print(gcd(14,63))
```

#### improving naive gcd

```python
def gcd(m,n):

    fm, fn = [],[]
    
    for i in range(1,max(m,n)+1):
        if m%i == 0:
            fm.append(i)
        
        if n%i == 0:
            fn.append(i)

    cf = []
    for f in fm:
        if f in fn:
            cf.append(f)
    
    return cf[-1]

print(gcd(14,63))
```

```python
def gcd(m,n):
    
    i = min(m,n)

    while i > 0:
        if m%i == 0 == n%i:
            return i
        else:
            i = i-1

print(gcd(14,63))
```

#### Euclid's algorithm for gcd

let\
m%d == n%d == 0\
m>n

m=ad , n=bd\
m-n = (a-b)d

gcd(m,n)= gcd(m-n)

```python
def gcd(m,n):
    
    # assume m >= n
    if m < n:
        m,n = n,m
    
    if m%n == 0:
        return n
    else:
        diff = m-n
        # diff > n? possible!
        return gcd(max(n,diff),min(n,diff))

print(gcd(14,63))
```

```python
def gcd(m,n):
    
    # assume m >= n
    if m < n:
        m,n = n,m
    
    while m%n != 0:
        diff = m-n
        # diff > n? possible!
        m,n= max(n,diff),min(n,diff)
    return n

print(gcd(14,63))
```

```python
def gcd(m,n):
    
    # assume m >= n
    if m < n:
        m,n = n,m
    
    if m%n == 0:
        return n
    else:
        return gcd(n,m%n) # m%n < n, always!

print(gcd(14,63))
```
