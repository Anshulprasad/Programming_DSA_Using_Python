sorted case

Binary Seach O(log(n))

```python
def binarySearch(seq,v,l,r):
    
    if l > r:
        return -1
    
    mid= (l + r)//2

    if seq[mid]==v:
        return mid
    
    elif seq[mid]<v:
        return binarySearch(seq,v,mid+1,r)
    
    else:
        return binarySearch(seq,v,l,mid)

l = [1,2,3,4,5,6]
print(binarySearch(l,5,0,4))
```
