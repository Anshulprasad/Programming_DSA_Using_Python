passing values to functions

```python
def update(l,i,v):
    if 0<=i<len(l):
        l[i]=v
        return True
    else:
        v = v+1
        return False

l = [1,2,3,4]
v = 8
i = 2

a = update(l,i,v)
print(a,l,v)

i=5

b = update(l,i,v)
print(b,l,v)
```

recursive function

```python
def factorial(x):
    if x<=1:
        return x
    else:
        return x*factorial(x-1)

print(factorial(4))
```

prime

```python
def factors(x):
    f = []
    for i in range(1,x+1):
        if x%i==0:
            f.append(i)
    return f

def isprime(n):
    return factors(n) == [1,n]

print(isprime(5))
```

first n primes

```python
def nprimes(n):
    count,i,plist = 0,1,[]

    while count<n:
        if isprime(i):
            plist.append(i)
            count = count + 1
        i = i+1

    return plist

print(nprimes(5))
```
