points on a plane

```python
from math import sqrt

class point:
    def __init__(self,a,b):
        self.x = a
        self.y = b
    
    def translate(self,deltax,deltay):
        self.x += deltax
        self.y += deltay
    
    def distance(self):
        return sqrt(self.x**2 + self.y**2)

p = point(4,0)
p.translate(2,15)
print(p.x,p.y,p.distance())
```

```python
from math import atan, cos, sin, pi

class point:
    def __init__(self,a=0,b=0):
        self.r = sqrt(a**2 + b**2)

        if a == 0:
            self.theta = pi/2
        else:
            self.theta = atan(b/a)

    def odistance(self):
        return self.r
    
    def translate(self,deltax,deltay):
        a = self.r*cos(self.theta) + deltax
        b = self.r*sin(self.theta) + deltay
        # print(a,b)

        self.r = sqrt(a**2 + b**2)
        self.theta = atan(b/a) if a!=0 else pi/2
    
p = point(4,0)
p.translate(2,15)
print(p.r*cos(p.theta),p.r*sin(p.theta),p.odistance(),p.theta)
```

```python
class point:

    def __init__(self,a):
        self.x =a

    def __str__(self):
        return ("__str__"+str(self.x))
    
    def __add__(self,a):
        return self.x * a.x
    
    def __mult__(self,a):
        return self.x + a.x + 50
    
    def __lt__(self, a): return '__lt__'
    def __gt__(self, a): return '__gt__'
    def __le__(self, a): return '__le__'
    def __ge__(self, a): return '__ge__'

p0,p1,p2=point(0), point(1),point(2)

print(p1.__str__())

print(p0 + p1)
print(p1.__add__(p2))

# print(p1 * p2)
print(p1.__mult__(p2))

print(p1 > p2)
print(p1 >= p2)
print(p1 < p2)
print(p1 <= p2)
```
