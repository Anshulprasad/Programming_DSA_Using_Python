opening a file

```python
fh = open("_temp1_.txt","r")
```

Read through file handle

```python
contents = fh.read()
print(contents)
print()

fh.seek(0)
contents = fh.readline()
print(contents)
print()

fh.seek(0)
contents = fh.readlines()
print(contents)
print()

fh.seek(3)
contents = fh.read(10)
print(contents)
```

End of file

```python
print(fh.read())
print(fh.readline())

# "" singal end of file
print("End of file")
print(fh.read())
print(fh.readline())
```

Writing to a file

```python
s = "Hello World!\nhow are you?\n"
fh = open("_temp2_.txt",'w')
fh.write(s)

l = ["1.","2.\n","3."]
fh.writelines(l)

fh.flush()
fh.close()
```

processing file line by line

```python
fh = open("_temp1_.txt",'r')

# fh.seek(0)
for l in fh.readlines():
    print(l,end='')
```

copying a file

```python
infile = open("_temp1_.txt",'r')
outfile = open("_temp2_.txt",'w')

for line in infile.readlines():
    outfile.write(line)

infile.close()
outfile.close()
```

```python
infile = open("_temp1_.txt",'r')
outfile = open("_temp2_.txt",'a')

contents = infile.readlines()
outfile.writelines(contents)

infile.close()
outfile.close()
```

strip new line character

```python
fh = open('_temp1_.txt','r')

contents = fh.readlines()

for line in contents:
    s = line[:-1]
    print(s,end='')
    
print("\n")
for line in contents:
    s = line.strip()
    print(s,end='')
```
