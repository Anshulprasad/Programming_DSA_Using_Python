nested lists

```python
nested = [[2,[37]], 4, ["hello"]]

print(nested[0][1][0])
print(nested[0][1])
print(nested[0][1:2])

print(nested[1:])
```

lists are mutable

```python
nested[0][1][0] = 7
print(nested)
```

for immutable values, we can assume that assignment makes a fresh copy of a value

```python
x= 5
y= x

print(id(x),id(y))

x= 7

print(x,y)
print(id(x),id(y))
```

for mutable values, assignment does not make a fresh copy

```python
list1 = [1,3,5,7]
list2= list1
list1[2] = 11

# list2 is just another name of list1. they are pointing to same list
print(list1,list2)
print(id(list1),id(list2))
```

copying lists

```python
list1 = [1,3,5,7]
list2= list1[:] # full slice
list1[2] = 11

print(list1,list2)
print(id(list1),id(list2))
```

digression on equality

```python
list1 = [1,3,5,7]
list2 = [1,3,5,7]
list3 = list2 # list1 and list2 refer to the same object

print(list1==list2)
print(list2==list3)
print(list1 is list2)
print(list2 is list3)
```

concatenation
