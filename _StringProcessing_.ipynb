Strip whitespace

```python
s = "  Hello world , I am good  "
print(s.rstrip())
print(s.lstrip())
print(s.strip())
```

searching for text

```python
print(s.find('world'))
print(s.find('world',2,6))

print(s.index('world'))
try:
    print(s.index('world',2,6))
except (SyntaxError,ValueError):
    print(-1)
```

search and replace

```python
print(s.replace('o','x'))

print(s.replace('o','x',2))
```

splitting a string

```python
print(s.split(' ',2))
```

joining strings

```python
a = s.split(' ')
print(a)

csvline = '|'.join(a)
print(csvline)
```

converting case

```python
print(s.capitalize())
print(s.lower())
print(s.upper())
print(s.title())
print(s.swapcase())
```

resizing strings

```python
print(s.center(50))
print(s.center(50,'*'))
print(s.ljust(50))
print(s.ljust(50,'*'))
print(s.rjust(50))
print(s.rjust(50,'*'))
```

other functions

```python
print('abc'.isalpha())
print('12'.isnumeric())
print('12abc'.isalnum())
```
