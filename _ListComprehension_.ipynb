## map

```python
def odd(n):
    return bool(n%2)

f= odd
l=[1,2,3,4,5]
print(list(map(f,l)))

for i in map(f,l):
    print(i,end=' ')
```

## filter

```python
def property(x):
    return x%2==0

l=[1,2,3,4,5]
f = property
print(list(filter(f,l)))

for i in filter(f,l):
    print(i,end=' ')
```

```python
def square(x):
    return x**2

def iseven(x):
    return x%2==0

l = map(square,filter(iseven,range(0,100)))
print(list(l))
```

## useful for initialising lists

```python
# (4 X 3) matrix
l = [[0 for i in range(3)]
     for j in range(4)]

print(l)

# each row in l points to same list zerolist
zerolist = [0 for i in range(3)]
l=[zerolist for j in range(4)]

l[1][1] = 7
l[0].extend([4,6])
print(l)
```
