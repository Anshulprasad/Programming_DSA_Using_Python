```python
l = []
if l:
    print("list")

s = ""
if s:
    print("string")

x = 0
if x:
    print("zero")
```

multiway branching, elif

```python
if 1:
    4
elif 2:
    8
else:
    5
```

loops: repeated actions

```python
for i in [1,2,3,4]:
    print(i)
```

```python
i = 4
while i>0:
    print(i)
    i = i-1
```
