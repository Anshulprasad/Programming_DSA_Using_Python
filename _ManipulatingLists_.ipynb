lists

```python
list1 = [1,3,5,6]
list2 = list1

list1 = list1[:2] + [7] + list1[3:]

print(list1)
print(list2)
```

extending a list

```python
list1 = [1,3,5,6]
list2 = list1

list1.append(12)

print(list1)
print(list2)
```

```python
list1 = [1,3,5,6]
list2 = list1

list1 = list1 + [12]

print(list1)
print(list2)
```

list functions

```python
list1 = [1,3,5,6]
list2 = list1

list1.append(12)
list1.extend(list2)
list1.remove(3)

print(list1)
```

further list manipulation

```python
list1 = [1,3,5,7]
list2 = list1

list1[1:3] = [11,34,54,123,234,2345]
print(list1,list2,sep='\n')
```

other functions

```python
list1 = [1,2,3,1,4,5]

list1.reverse()
print(list1)

print(list1.index(4))

list1.sort()
print(list1)
```
