string format() method

```python
s = "first: {1}, second: {0}".format(47,11)
print(s)

s = "first: {f}, second: {s}".format(s=11,f=47)
print(s)
```

now, real formatting

```python
s = "value:{0:3d}".format(4)
print(s)

s = "value:{0:9.4f}".format(47.523)
print(s)
```
