## Selection Sort
o(n^2)

```python
def SelectionSort(seq):

    for start in range(len(seq)):
        minPos=start
        for i in range(start,len(seq)):
            if seq[i] < seq[minPos]:
                minPos = i
        
        seq[minPos],seq[start]= seq[start],seq[minPos]
    return seq

SelectionSort([1,5,3,1234,345,-1])
```

## Insertion Sort
o(n^2)\
usually better than selection sort

```python
def InsertionSort(seq):

    for sliceEnd in range(len(seq)):
        pos = sliceEnd
        while pos>0 and seq[pos-1] > seq[pos]:
            seq[pos-1], seq[pos] = seq[pos], seq[pos-1]
            pos = pos - 1
    return seq

InsertionSort([1,5,3,1234,345,-1])
```
