```python
def explore(s,t,m,n):
    (sx,sy),(tx,ty) = s,t

    marked = [[0 for i in range(n)]
              for j in range(m)]
    
    marked[sx][sy] = 1
    queue = [(sx,sy)]

    while queue != []:
        (ax,ay) = queue.pop()

        for i,j in neighbors(ax,ay,m,n,marked):
            # print(i,j)

            if marked[i][j] == 0:
                marked[i][j] = 1
                queue.insert(0,(i,j))

    print(marked)

    return marked[tx][ty]

def neighbors(ax,ay,m,n,marked):
    a = [(ax-1,ay+2),(ax+1,ay+2),(ax-1,ay-2),(ax+1,ay-2),(ax-2,ay+1),(ax-2,ay-1),(ax+2,ay+1),(ax+2,ay-1)]
    
    l=[]

    for i,j in a:
        if 0<=i<m and 0<=j<n and marked[i][j]==0:
            l.append((i,j))

    print(l)
    return l

print(explore((4,2),(7,7),8,10))
```
