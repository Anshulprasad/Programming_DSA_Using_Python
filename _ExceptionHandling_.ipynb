## SytaxError

```python
try:
    l = [0,1;2]
except SyntaxError:
    print("Error!")
else:
    print("NoError!")
```

## ZeroDivisionError

```python
try:
    a= 0
    y= 5/a
except:
    print("Error!")
else:
    print("NoError!")
```

## ValueError

```python
s = '124af'
try:
    a=int(s)
except ValueError:
    print("ValueError")
else:
    print("NoError")
```

## FileNotFoundError

```python
try:
    fh = open('x.py','r')

except FileNotFoundError:
    print("File does not exist")
    
else:
    print("NoError")
```

## IndexError

```python
try:
    l=[1,2]
    l[3]=5
except IndexError:
    print("Error!")
else:
    print("NoError!")
```

## KeyError

```python
scores={'kohli':[1,2],'rohit':[3,4]}

name = 'dhoni'
s = 100
try:
    scores[name].append(s)
except KeyError:
    scores[name] = [s]

print(scores)
```
